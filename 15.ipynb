Introducción y ConceptoOrden del Caos: Representación visual del proceso de estructurar datos crudos e inconexos en modelos explicativos organizados.   La metáfora del pastel: "El aprendizaje no supervisado sería el pastel, el supervisado el glaseado...". Resalta que la mayor parte de la información y la estructura base proviene de datos sin etiquetar.   Módulo 1: Datos No Supervisados y AgrupamientoDefinición de Agrupamiento: Consiste en identificar y asignar datos similares a grupos definidos basándose en sus características intrínsecas (ej. agrupar figuras geométricamente similares o elementos botánicos).   Casos de Uso Prácticos:Segmentación de clientes/usuarios: Agrupación de perfiles para análisis de audiencia.   Sistemas de recomendación: Agrupación de hábitos de consumo en plataformas digitales (móviles/video).   Búsqueda y extracción de patrones: Identificación de subconjuntos específicos mediante inspección visual o analítica.   Módulo 2: Agrupamiento y K-Means (Agrupando a Gran Escala)Algoritmo K-Means: Diseñado para estructurar y clasificar grandes volúmenes de datos. Identifica eficazmente múltiples grupos/clusteres continuos.   Evaluación del modelo: Muestra la importancia de equilibrar el costo/complejidad (C) con el espacio/simetría (S) al formar las particiones.   Módulo 3: Segmentación y Datos SemisupervisadosUso combinado de datos: Combina pequeñas cantidades de datos etiquetados con grandes volúmenes de datos no etiquetados para mejorar la precisión del modelo.   Aprendizaje Activo (Active Learning): Proceso iterativo de tres pasos:   Entrenar: El modelo identifica e aísla las instancias con mayor grado de incertidumbre o duda.   Etiquetar: Un experto humano revisa y asigna las etiquetas correspondientes a dichos casos críticos.   Iterar: Se repite el flujo para optimizar progresivamente el rendimiento general del sistema.   Módulo 4: DBSCAN, Formas y Estimación de DensidadAlgoritmo DBSCAN: Algoritmo basado en densidad capaz de identificar clusters de formas complejas, arbitrarias y continuas (ej. figuras irregulares o concéntricas) que algoritmos tradicionales como K-Means no logran capturar correctamente.   Módulo 5: Anomalías y NovedadesDetección de Anomalías: Identificación de patrones atípicos o fallas. Se ilustran dos enfoques:   Detección de fallas industriales: Identificación de piezas defectuosas en líneas de ensamblaje.   Detección de novedades: Identificación de elementos fuera del estándar habitual dentro de un conjunto homogéneo.   Módulo 6: Mezclas Gaussianas (GMM) y Criterios del ModeloModelos de Mezcla Gaussiana: Técnica probabilística que delimita áreas mediante contornos de densidad para aislar anomalías y puntos fuera de distribución (outliers).   Métricas de selección (BIC y AIC): Criterios analíticos (Bayesian Information Criterion y Akaike Information Criterion) utilizados para calcular de forma matemática el número óptimo de grupos dentro de modelos probabilísticos.

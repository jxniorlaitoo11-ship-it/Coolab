Biología a Redes Neuronal ArtificialesLas Redes Neuronales Artificiales (RNA) se inspiran en la estructura y funcionamiento del cerebro humano.En la biología:Las dendritas reciben señales químicas u eléctricas de otras neuronas.El cuerpo celular (soma) acumula y procesa la información entrante.El axón transmite la respuesta generada.La sinapsis es el punto de conexión que regula la fuerza de la señal transmitida a la siguiente neurona.En la versión artificial:Las entradas ($x$) representan los datos entrantes (análogo a las dendritas).Los pesos ($w$) determinan la relevancia o fuerza de cada entrada (análogo a las sinapsis).La suma ponderada más un sesgo ($b$) combina las señales (análogo al procesamiento en el soma).La función de activación determina si la neurona emite una respuesta y cuál es su magnitud (análogo a la transmisión del axón).02. El PerceptrónDesarrollado por Frank Rosenblatt, el perceptrón es el modelo de neurona artificial más simple y la base histórica del aprendizaje profundo.Funcionamiento: Toma múltiples entradas binarias o continuas, las multiplica por sus respectivos pesos, les suma un término de sesgo ($b$) y pasa el valor por una función de paso o escalón. Si el resultado supera cierto umbral, la salida es 1; de lo contrario, es 0.Ecuación fundamental: $\hat{y} = f(\sum w_i x_i + b)$Limitación principal: Un perceptrón simple solo puede clasificar datos que sean linealmente separables (capaz de dividirse con una sola línea o hiperplano). Es capaz de resolver funciones lógicas como AND u OR, pero falla completamente en problemas no lineales como la función XOR.03. MLP y Retropropagación (Backpropagation)Para resolver problemas complejos y no lineales se combinan múltiples perceptrones formando una red más amplia.Perceptrón Multicapa (MLP): Es una arquitectura compuesta por una capa de entrada, una o más capas ocultas y una capa de salida. Al utilizar funciones de activación no lineales en las capas ocultas (como ReLU, Sigmoid o Tanh), el MLP adquiere la capacidad de aprender relaciones y fronteras de decisión no lineales complejas.Algoritmo de Retropropagación (Backpropagation): Es el proceso matemático central mediante el cual la red aprende:Forward Pass (Paso adelante): Los datos avanzan capa por capa produciendo una predicción.Cálculo de la pérdida: Se compara la predicción con el valor real usando una función de costo o pérdida (Loss Function).Backward Pass (Paso atrás): Aplicando la regla de la cadena del cálculo diferencial, se calcula la contribución de cada peso al error total (el gradiente del error).Optimización: Algoritmos de optimización (como Descenso de Gradiente o Adam) ajustan ligeramente los pesos en la dirección opuesta al gradiente para reducir la pérdida en la siguiente iteración.04. API Secuencial de KerasLa API Secuencial es el enfoque más simple y directo para construir modelos en Keras.Características:Permite crear modelos capa por capa de forma lineal, donde cada capa tiene exactamente un tensor de entrada y un tensor de salida.No permite bifurcaciones, conexiones cruzadas ni múltiples entradas o salidas.Casos de uso ideal: Prototipado rápido, redes feedforward tradicionales, clasificadores sencillos y arquitecturas estándar donde la información fluye en una sola dirección.05. La API Funcional de KerasLa API Funcional trata a las capas como funciones matemáticas explícitas que reciben tensores y devuelven tensores.Características:Permite crear grafos de capas completamente arbitrarios y no lineales.Facilita la creación de arquitecturas con múltiples entradas (por ejemplo, texto e imágenes simultáneamente), múltiples salidas o ramas paralelas.Soporta conexiones residuales o "skip connections" (donde la salida de una capa anterior se suma directamente a una capa posterior, como en las redes ResNet).Casos de uso ideal: Arquitecturas complejas, redes con bloques de conexiones compartidas y la mayoría de los modelos de producción avanzados.06. API de Subclases de Keras (Model Subclassing)La API de Subclases ofrece el nivel más alto de flexibilidad y control al utilizar un estilo de programación orientada a objetos puro en Python.Características:Se define una clase personalizada que hereda de la clase base Model de Keras.Las capas e hiperparámetros se instancian en el método inicializador __init__.El flujo de trabajo y la lógica de propagación hacia adelante se implementan manualmente dentro del método call.Permite incorporar lógica imperativa en Python, como bucles for o condicionales if que alteren el flujo de datos dinámicamente según la entrada.Casos de uso ideal: Investigación avanzada, implementación de arquitecturas novedosas que no se adaptan al flujo estático de la API funcional, creación de capas personalizadas y entrenamiento con bucles de control a bajo nivel.
